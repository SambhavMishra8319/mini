# Exploratory Data Analysis
Mental Health Risk Prediction — dataset exploration.
Run from the project root, or adjust the path below.

```python
import sys, os
sys.path.insert(0, os.path.abspath('..'))
import pandas as pd
from src import config
from src.preprocessing import preprocess_series

path = config.DATASET_PATH if os.path.exists(config.DATASET_PATH) else config.SAMPLE_DATASET_PATH
df = pd.read_csv(path)
print('Loaded:', path)
df.head()
```

```python
print('Rows:', len(df))
print('Missing values:\n', df.isna().sum())
print('Duplicate texts:', df[config.TEXT_COLUMN].duplicated().sum())
df[config.TARGET_COLUMN].value_counts()
```

```python
df['clean_text'] = preprocess_series(df[config.TEXT_COLUMN])
df['word_count'] = df['clean_text'].str.split().apply(len)
df['word_count'].describe()
```

```python
import matplotlib.pyplot as plt
df[config.TARGET_COLUMN].value_counts().plot(kind='bar', title='Class Distribution')
plt.tight_layout()
plt.show()
```
